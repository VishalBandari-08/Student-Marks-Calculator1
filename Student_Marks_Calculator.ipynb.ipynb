```python
print("STUDENT MARKS CALCULATOR")

name = input("Enter student name: ")

marks = []

for i in range(1, 6):
    while True:
        mark = int(input(f"Enter marks in Subject {i} (0-100): "))

        if 0 <= mark <= 100:
            marks.append(mark)
            break
        else:
            print("Invalid marks! Enter marks between 0 and 100.")

total = sum(marks)
percentage = total / 5

print("\n----- STUDENT RESULT -----")
print("Student Name:", name)
print("Total Marks:", total, "/ 500")
print("Percentage:", percentage, "%")

if min(marks) < 35:
    print("Grade: F")
    print("Result: Fail")
elif percentage >= 90:
    print("Grade: A")
    print("Result: Pass")
elif percentage >= 75:
    print("Grade: B")
    print("Result: Pass")
elif percentage >= 60:
    print("Grade: C")
    print("Result: Pass")
elif percentage >= 40:
    print("Grade: D")
    print("Result: Pass")
else:
    print("Grade: F")
    print("Result: Fail")
```

```python
import pandas as pd
from IPython.display import display

student_name = "Vishal"

subjects = [
    "Subject 1",
    "Subject 2",
    "Subject 3",
    "Subject 4",
    "Subject 5"
]

marks = [85, 90, 80, 75, 90]

total = sum(marks)
percentage = total / len(marks)

if min(marks) < 35:
    grade = "F"
    result = "Fail"
elif percentage >= 90:
    grade = "A"
    result = "Pass"
elif percentage >= 75:
    grade = "B"
    result = "Pass"
elif percentage >= 60:
    grade = "C"
    result = "Pass"
elif percentage >= 40:
    grade = "D"
    result = "Pass"
else:
    grade = "F"
    result = "Fail"

data = {
    "Subject": subjects,
    "Marks": marks,
    "Maximum Marks": [100] * 5
}

df = pd.DataFrame(data)

print("STUDENT MARKS REPORT")
print("Student Name:", student_name)
display(df)

print("Total Marks:", total, "/ 500")
print("Percentage:", percentage, "%")
print("Grade:", grade)
print("Result:", result)
```

```python
print("AI STUDENT PERFORMANCE ANALYSIS")

subjects = [
    "Subject 1",
    "Subject 2",
    "Subject 3",
    "Subject 4",
    "Subject 5"
]

marks = [85, 90, 80, 75, 90]

average = sum(marks) / len(marks)

print("Average Marks:", average)

if average >= 85:
    print("Performance: Excellent")
    print("Suggestion: Keep maintaining your performance.")

elif average >= 70:
    print("Performance: Good")
    print("Suggestion: Practice regularly to improve your marks.")

elif average >= 50:
    print("Performance: Average")
    print("Suggestion: Spend more time on difficult subjects.")

else:
    print("Performance: Needs Improvement")
    print("Suggestion: Follow a daily study plan.")

lowest = min(marks)
index = marks.index(lowest)

print("Subject needing improvement:", subjects[index])
print("Marks in that subject:", lowest)
```

```python
print("STUDENT PERFORMANCE COMPARISON")

student_name = "Vishal"
student_marks = [85, 90, 80, 75, 90]

# Example marks of other students
other_students = [
    [70, 75, 80, 65, 85],
    [90, 85, 88, 92, 95],
    [60, 70, 65, 75, 80]
]

student_average = sum(student_marks) / len(student_marks)

class_averages = [
    sum(marks) / len(marks)
    for marks in other_students
]

class_average = (
    student_average + sum(class_averages)
) / (len(class_averages) + 1)

print("Student Name:", student_name)
print("Student Average:", student_average)
print("Class Average:", round(class_average, 2))

if student_average > class_average:
    print("Performance: Above class average")
elif student_average < class_average:
    print("Performance: Below class average")
else:
    print("Performance: Equal to class average")
```

```python
import gradio as gr

def calculate_result(name, m1, m2, m3, m4, m5):
    marks = [m1, m2, m3, m4, m5]

    # Validate marks
    if not name.strip():
        return "Please enter the student's name."

    if any(m is None or m < 0 or m > 100 or not float(m).is_integer()
           for m in marks):
        return "Enter whole-number marks between 0 and 100."

    marks = [int(m) for m in marks]
    total = sum(marks)
    percentage = total / 5

    if min(marks) < 35:
        grade = "F"
        result = "Fail"
    elif percentage >= 90:
        grade = "A"
        result = "Pass"
    elif percentage >= 75:
        grade = "B"
        result = "Pass"
    elif percentage >= 60:
        grade = "C"
        result = "Pass"
    elif percentage >= 40:
        grade = "D"
        result = "Pass"
    else:
        grade = "F"
        result = "Fail"

    return (
        f"Student Name: {name}\n"
        f"Subject Marks: {marks}\n"
        f"Total Marks: {total}/500\n"
        f"Percentage: {percentage:.2f}%\n"
        f"Grade: {grade}\n"
        f"Result: {result}"
    )

app = gr.Interface(
    fn=calculate_result,
    inputs=[
        gr.Textbox(label="Student Name"),
        gr.Number(label="Subject 1 Marks", minimum=0, maximum=100),
        gr.Number(label="Subject 2 Marks", minimum=0, maximum=100),
        gr.Number(label="Subject 3 Marks", minimum=0, maximum=100),
        gr.Number(label="Subject 4 Marks", minimum=0, maximum=100),
        gr.Number(label="Subject 5 Marks", minimum=0, maximum=100)
    ],
    outputs=gr.Textbox(label="Student Result"),
    title="Student Marks Calculator",
    description="Enter student details and marks to calculate the result."
)

app.launch(share=True)
```
